```python
import tensorflow as tf
```

```python
# Load images
data = tf.keras.utils.image_dataset_from_directory(
    r"c:\Users\INTEL AI DATA LAB\Downloads\cap_3\jai.python\Country Flag.Dataset\Pollution.dataset\Train",
    image_size=(64, 64),
    batch_size=2
)
```

```python
# CNN model
model = tf.keras.Sequential([
    tf.keras.layers.Rescaling(1./255),
    tf.keras.layers.Conv2D(16, (3,3), activation="relu"),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(3, activation="softmax")
])
# Compile
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
# Train
model.fit(data, epochs=50)
model.summary()
```

```python
# Prediction
img = tf.keras.utils.load_img(r"c:\Users\INTEL AI DATA LAB\Downloads\cap_3\jai.python\Country Flag.Dataset\Pollution.dataset\Test\Air pollution\2.jpg", 
                              target_size=(64,64))
x = tf.keras.utils.img_to_array(img)
x = tf.expand_dims(x, 0)

p = model.predict(x)
print(data.class_names[p.argmax()])
```
